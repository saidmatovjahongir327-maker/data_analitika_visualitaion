```python
import pandas as pd
df = pd.read_csv('GooglePlayStore_cleaned.csv')
df.info()
```

```python
df['size(kb)'].plot(kind='hist' , bins = 5)
```

```python
df[df['type'] == 'paid']['price'].plot(kind='hist')
```

```python
df.plot(x='rating', y = 'installs', kind = 'scatter')
```

```python
df['category'].value_counts().plot(kind = 'pie')
```

```python
df['category'].value_counts().plot(kind = 'bar')
```

```python
d1  = df[df['type'] == 'free'].groupby('content_rating')['installs'].mean()
d1.plot(kind = 'barh')
```

```python
d = df.pivot_table(index = 'content_rating',columns='type',values='installs',aggfunc='mean')
d.plot(kind = 'barh')
```

```python
d = df.pivot_table(index = 'content_rating',columns='type',values='installs',aggfunc='mean')
d.plot(kind = 'barh',subplots = True)
```

```python
d = df.pivot_table(index = 'content_rating',columns='type',values='installs',aggfunc='mean')
d.plot(kind = 'barh',grid =True)
```

```python
d = df.pivot_table(index = 'content_rating',columns='type',values='installs',aggfunc='mean')
d.plot(kind = 'barh',subplots = True , layout = (1,2))
```

```python
d = df[df['type'] == 'paid'].groupby('content_rating')['installs'].mean()
d.plot(kind = 'line')
```
